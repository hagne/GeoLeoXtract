# Download

```python
import os
import requests
from bs4 import BeautifulSoup
```

```python
# Base URL of the directory
BASE_URL = "https://www.star.nesdis.noaa.gov/pub/smcd/hzhang/TEMPO/AOD_ALH/"
LOCAL_DIR = "./TEMPO_AOD_ALH/"  # Local directory to save files
LOCAL_DIR = "/nfs/stu3data2/Satellite_data/tempo/AOD_ALH/original/"
```

```python
def download_file(url, local_path):
    """Download a single file."""
    response = requests.get(url, stream=True)
    if response.status_code == 200:
        with open(local_path, 'wb') as file:
            for chunk in response.iter_content(chunk_size=1024):
                file.write(chunk)
        print(f"Downloaded: {url}")
    else:
        print(f"Failed to download: {url}")

def scrape_and_download(base_url, local_dir):
    """Recursively scrape and download files and subdirectories."""
    # Get the HTML content of the directory
    response = requests.get(base_url)
    if response.status_code != 200:
        print(f"Failed to access {base_url}")
        return

    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Create local directory if it doesn't exist
    os.makedirs(local_dir, exist_ok=True)

    # Find all links in the HTML
    for link in soup.find_all('a', href=True):
        href = link['href']
        if href == "../":  # Skip parent directory link
            continue

        # If it's a directory (ending with '/')
        if href.endswith('/'):
            subdir_url = base_url + href
            subdir_local = os.path.join(local_dir, href.strip('/'))
            scrape_and_download(subdir_url, subdir_local)
        else:
            # It's a file, download it
            file_url = base_url + href
            file_local = os.path.join(local_dir, href)
            download_file(file_url, file_local)
```

```python
# Start the process
scrape_and_download(BASE_URL, LOCAL_DIR)
```
