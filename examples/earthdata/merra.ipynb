```python
sites = atmsrf.network.stations

data_product = 'M2T1NXAER'
# p2fld_out = pl.Path('/home/hagen/tmp/M2T1NXAER/')

# p2fld_out = pl.Path('/home/grad/htelg/tmp/M2T1NXAER/projected')
p2fld_out = pl.Path('/nfs/stu3data2/Model_data/merra_2/M2T1NXAER/projected2surfrad/')
p2fld_orig = pl.Path('/home/grad/htelg/tmp/M2T1NXAER/orig')
p2fld_orig.mkdir(exist_ok=True)

```

```python
scraper = glx.scrapers.earthdata.CMRSraperGlobal
scraper_kwargs = dict(product=data_product,
                        p2fld_orig = p2fld_orig,
                        reporter=None,
                        overwrite=True, # True to make sure incomplete files are getting redownloaded
                        verbose=True,
                        # test = 'full',
                     )

product = glx.merra.make_product_v01
product_kwargs = dict()

self = glx.projector.prolab.ChunckProcessor(
                                            start='20230101',
                                            end='20240101',
                                            chunksize='MS',
                                            sites=sites,
                                            p2fld_out=p2fld_out,
                                            name_addon='projection2surfrad',
                                            scraper=scraper,
                                            scraper_kwargs=scraper_kwargs,
                                            product=product,
                                            product_kwargs=product_kwargs,
                                            # overwrite=True,
                                            # test='full',
                                            remove_originals = True,
                                            )
```
